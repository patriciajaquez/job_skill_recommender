# 🏗️ Phase 3: Data Pipeline Architecture Planning

## 📊 **Data Architecture Design**

### **Data Flow Architecture:**

```
📡 APIs → 🔄 N8N → 🧠 ML Processing → 💾 Azure SQL → 📱 Streamlit App
                     ↓
                  🛡️ Backup Generation → 📦 Azure Blob Storage
```

### **Database Schema Design:**

**Core Tables:**
1. **`jobs`** - Main job postings data
2. **`skills`** - Extracted and standardized skills
3. **`job_skills`** - Many-to-many relationship
4. **`companies`** - Company information
5. **`locations`** - Geographic data
6. **`trends`** - Market trend analytics
7. **`data_quality`** - Data quality metrics
8. **`api_health`** - API status monitoring

````python
# Database Schema Definition
database_schema = {
    "jobs": {
        "columns": {
            "id": "UUID PRIMARY KEY",
            "title": "VARCHAR(500) NOT NULL",
            "description": "TEXT",
            "company_id": "UUID FOREIGN KEY",
            "location_id": "UUID FOREIGN KEY",
            "salary_min": "DECIMAL(10,2)",
            "salary_max": "DECIMAL(10,2)",
            "currency": "VARCHAR(3)",
            "employment_type": "VARCHAR(50)",
            "experience_level": "VARCHAR(50)",
            "remote_allowed": "BOOLEAN",
            "posted_date": "DATETIME",
            "source_api": "VARCHAR(50)",
            "external_id": "VARCHAR(255)",
            "url": "TEXT",
            "created_at": "DATETIME DEFAULT CURRENT_TIMESTAMP",
            "updated_at": "DATETIME DEFAULT CURRENT_TIMESTAMP"
        },
        "indexes": ["title", "location_id", "posted_date", "source_api"]
    },
    "skills": {
        "columns": {
            "id": "UUID PRIMARY KEY",
            "name": "VARCHAR(255) UNIQUE NOT NULL",
            "category": "VARCHAR(100)",
            "popularity_score": "DECIMAL(5,2)",
            "demand_trend": "VARCHAR(20)",
            "created_at": "DATETIME DEFAULT CURRENT_TIMESTAMP"
        }
    },
    "job_skills": {
        "columns": {
            "job_id": "UUID FOREIGN KEY",
            "skill_id": "UUID FOREIGN KEY",
            "importance_score": "DECIMAL(3,2)",
            "extraction_confidence": "DECIMAL(3,2)"
        },
        "primary_key": "(job_id, skill_id)"
    }
}

print("📊 Database Schema Design:")
for table, config in database_schema.items():
    print(f"\n🏷️ Table: {table}")
    print(f"Columns: {len(config['columns'])}")
    if 'indexes' in config:
        print(f"Indexes: {', '.join(config['indexes'])}")
````
